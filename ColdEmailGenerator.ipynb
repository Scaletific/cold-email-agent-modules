```python

#Business Analyst
Analyst = Agent (
    role = "Business Analyst",
    goal = "Identify the subniche within {company} specific to its industry",
    backstory = "You are a business analyst and your primary function is to identify specific subniches,"
    "of {company} and the broader industry at large."
    " With this information, you aid businesses in targeting precise market segments. You collect information from {company}"
    "{industry}, and {briefDes}. Armed with these details, conduct thorough research on the current subniches within{industry}",
    allow_delegation= False,
    verbose = True
)
```

```python
#Job Title Agent
Profiler = Agent(
    role = "Business Portfolio Analyst",
    goal = "Identify and list the Job Titles within {company} that are responsible for or will benefit from{Offer},"
    "and the supervisor in charge of the Job Titles.",
    backstory = " You are a Business Portfolio Analyst that identifies relevant job titltes within {company} and {industry},"
    "that would benefit from {Offer}. You analyse information from Business Analyst and based on its output, make this decision."
    "Your mission is to enhance cold campaigns by targeting Job Titles and supervisor to these Job titles that are the final decision makers.",
    allow_delegation= False,
    verbose = True
)
```

```python
#Pain points Analyst
Pain_points_analyst = Agent(
    role = "Business Pain Points Analyst",
    goal = "Identify pain points of Job Titles provided by the Business Portfolio Analyst, ranking them from the one that contributes to loss of revenue.",
    backstory = " You are Business Pain Points Analyst that identifies the key pain points of Job titles provided by Business Portfolio Analyst,"
    "ranking the key pain points in order of the intensity of the impact on revenue."
    "You also filter the pain points and identify the solutions that were used to address them,"
    "in the past."
    "You based your ranking based on reviews and setiments from the targeted Job Titles within {industry}.",
    allow_delegation= False,
    verbose = True

)
```

```python
#Cold  Email Generator
Cold_email_generator = Agent(
    role = "Cold Email Generator",
    goal = " Generate cold emails in less than 100 words for the various Job Titles provided by Business Portfolio Analyst.",
    backstory = "You are a cold email writer that takes into account the information and details provided by the Business Analyst, the Business Portfolio Analyst,"
    "and the Business Pain Points Analyst. Write a non salesy cold email that addresses the Job titles that struggled with the pain ponts identified by the Business Pain Points Analyst. "
    "Mention that our {sender} has specifically  helped a similar company  that struggled with the same problem in the past."
    "Include in each mail how a bullet of the pain points trying to address, the industry and modern solutions to address the pain point"
    "Using this as a template:"
    "Pain point: "
    "Industry:"
    "Modern Solution:"
    "Cold Email:",
    allow_delegation= False,
    verbose = True
)
```

```python
#Tasks
subniche =  Task(
    description = (
        "1. Identify the subniche within {company} specific to its industry.\n"
        "2. Using information provided by business analyst, identify specific subniches,"
    "of {company} and the broader industry at large.\n"
    ),
    expected_output = "A comprehensive detail of the subniches within {company} and the broader industry at large.",
    agent= Analyst,
)
```

```python
#Tasks
profile = Task(
    description = (
        "1. Identify and list the Job Titles within {company} that are responsible for or will benefit from {Offer}.\n"
        "2. Using the information provided by Business Portfolio Analyst, identify the Job Titles' supervisors of the subniches.\n"
    ),
    expected_output = "A comprehensive detail of the Job Titles' supervisors of the subniches that will benefit from {Offer}",
    agent= Profiler,
)
```

```python
#Tasks
painPoints = Task(
    description =(
        "1. Identify pain points of Job Titles provided by the Business Portfolio Analyst.\n"
        "2. These pain points should be ranked in order of the intensity of the impact on revenue.\n"
        "3. Filter the pain points and identify the solutions that were used to address them.\n"
        "4. You based your ranking based on reviews and setiments from the targeted Job Titles within {industry}.\n"
    ),
    expected_output = "A comprehensive detail of the pain points of Job Titles provided by the Business Portfolio Analyst.",
    agent= Pain_points_analyst,
)
```

```python
#Tasks
coldEmail = Task(
    description = (
        "1. Generate cold emails in less than 100 words for the various Job Titles provided by Business Portfolio Analyst. Explain at the end each email how,"
        " you came by came by these emails, which should include the processes and the steps taken."
        "2. Mention that our {sender} has specifically  helped a similar company  that struggled with the same problem in the past.\n"
        "3. Additionally include how {sender} can address their pain points with modern solutions"
        "4. You are a cold email writer that takes into account the information and details provided by the Business Analyst, the Business Portfolio Analyst,"
    "and the Business Pain Points Analyst.\n"
        ),
    expected_output = "A less than 100 words cold email for the various Job Titles provided by Business Portfolio Analyst.",
    agent= Cold_email_generator,
)
```

```python
#Crew
crew = Crew(
    agents = [Analyst, Profiler, Pain_points_analyst, Cold_email_generator],
    tasks = [subniche, profile, painPoints, coldEmail],
    verbose = 2
)
```
